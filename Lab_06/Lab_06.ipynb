### Assignment 4 prerequisite
Paste your completed `run_one_trial(N)` function from **Assignment 4, Exercise 6** below. It should return `(chosen_name, chosen_skill)`. The trial simulation and benchmark strategy were developed in Assignment 4, Exercises 6–8.

````python
# Paste your run_one_trial(N) function from Assignment 4 here.
def run_one_trial(N):
    # Shuffle interview order
    interview_order = list(lab_partners.keys())
    random.shuffle(interview_order)

    # Find benchmark for first N
    best_of_first_N = 0
    for i in range(N):
        temp_skill = lab_partners[interview_order[i]]["lab_skills"]
        if(temp_skill > best_of_first_N):  
            best_of_first_N = temp_skill

    # First to beat
    chosen_name = interview_order[-1]
    chosen_skill = lab_partners[chosen_name]["lab_skills"]

    for i in range(N, len(interview_order)):
        temp_skill = lab_partners[interview_order[i]]["lab_skills"]
        if(temp_skill > best_of_first_N):  
            chosen_name = interview_order[i]
            chosen_skill = temp_skill
            break

    return chosen_name, chosen_skill
````
